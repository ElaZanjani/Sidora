```python
!pip -q install librosa
from google.colab import drive
drive.mount('/content/drive')
import glob, os, librosa, librosa.display, numpy as np, matplotlib.pyplot as plt

kok = '/content/drive/MyDrive/sidora_veri'  # klasör konumu
siniflar = ['imdat', 'yardim_et', 'bagirma_cigilik', 'normal_konusma', 'ortam_sesi']
var = [s for s in siniflar if glob.glob(f'{kok}/{s}/*')]  # boş olanları atlar

fig, axes = plt.subplots(len(var), 1, figsize=(8, 3 * len(var)))
axes = np.atleast_1d(axes)

for ax, s in zip(axes, var):
    dosya = sorted(glob.glob(f'{kok}/{s}/*'))[0]
    y, sr = librosa.load(dosya, sr=16000, mono=True, offset=1, duration=4)
    mel = librosa.power_to_db(librosa.feature.melspectrogram(y=y, sr=sr, n_mels=64), ref=np.max)
    librosa.display.specshow(mel, sr=sr, x_axis='time', y_axis='mel', ax=ax)
    ax.set_title(f'{s} ({os.path.basename(dosya)})')

plt.tight_layout()
plt.show()
```
